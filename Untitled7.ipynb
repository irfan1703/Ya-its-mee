```python
import nltk
from transformers import AutoTokenizer

# 1. Download NLTK packages
try:
    nltk.data.find('corpora/stopwords')
except LookupError:
    nltk.download('stopwords')

try:
    nltk.data.find('tokenizers/punkt')
except LookupError:
    nltk.download('punkt')

print("NLTK 'stopwords' and 'punkt' packages are downloaded.")

# 2. Define the string to be tokenized
text = "This is an example sentence to demonstrate tokenization using transformers."

# 3. Initialize a tokenizer from the transformers package
# You can choose a pre-trained model. 'bert-base-uncased' is a good general choice.
tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")

# 4. Tokenize the string
tokens = tokenizer.tokenize(text)

# 5. Print the original string and the tokens
print(f"\nOriginal String: {text}")
print(f"Tokens (using transformers): {tokens}")

# If you want to see the token IDs as well (useful for model input)
# input_ids = tokenizer.encode(text, add_special_tokens=True)
# print(f"Token IDs (using transformers): {input_ids}")
# print(f"Decoded tokens from IDs: {tokenizer.convert_ids_to_tokens(input_ids)}")
```

```python
import re

def tokenize_specific_sentence(sentence):

    # Replace smart quote with a straight quote for consistent handling
    sentence = sentence.replace('’', "'")


    pattern = re.compile(r"(\s+|'|\.|\(|\)|\?|:|!|,)")

    # Use re.split to split the sentence by the pattern.
    # This returns a list of words and the captured delimiters.
    # We filter out any empty strings that might result from the split.
    tokens = [token for token in pattern.split(sentence) if token]

    return tokens

# The sentence to tokenize as per the user's request.
target_sentence = "Find out the words words that don’t usually appear in a standard lexicon? The separators are: whitespaces, quote (’), fullstop/period (.), parenthesis, are kept as tokens, tokenize the earlier sentence."

tokenized_output_specific = tokenize_specific_sentence(target_sentence)

print("--- Python Program Output ---")
print("Original Sentence:\n", target_sentence)
print("\nTokens:\n", tokenized_output_specific)

# Identifying words/tokens that might not typically appear in a standard lexicon
# from THIS tokenized output, based on our previous general analysis.
non_lexicon_candidates = set()

for token in tokenized_output_specific:
    # Check for parts of contractions like "'t"
    if token.startswith("'") and len(token) > 1:
        non_lexicon_candidates.add(token)
    # Check for the apostrophe itself, which is a token here
    elif token == "'":
        non_lexicon_candidates.add(token)
    # Other punctuation marks are also tokens, but not "words" in a lexicon.
    elif token in ['.', '(', ')', '?', ':', ',', '!', ' ']: # Added ' ' for whitespace token
        non_lexicon_candidates.add(token)

print("\n--- Words/Tokens that might not typically appear in a standard lexicon (from the tokenized sentence) ---")
# Convert the set to a list for consistent output.
# Sort for stable output, though not strictly required.
print(sorted(list(non_lexicon_candidates)))
```

```python
import numpy as np

def levenshtein_distance_chart(word1, word2):
    """
    Calculates the Levenshtein (edit) distance between two words
    using a chart-based (dynamic programming) algorithm.

    Args:
        word1 (str): The first word.
        word2 (str): The second word.

    Returns:
        tuple: A tuple containing:
            - int: The Levenshtein distance.
            - numpy.ndarray: The filled dynamic programming table (chart).
    """
    n = len(word1)
    m = len(word2)

    # Initialize the DP table (chart) with dimensions (n+1) x (m+1)
    # The +1 accounts for the empty string prefixes.
    dp_table = np.zeros((n + 1, m + 1), dtype=int)

    # Initialize the first row and first column
    # dp_table[i][0] represents the cost of transforming an empty string
    # into the first 'i' characters of word1 (i.e., i deletions).
    for i in range(n + 1):
        dp_table[i][0] = i
    # dp_table[0][j] represents the cost of transforming the first 'j' characters
    # of word2 into an empty string (i.e., j insertions).
    for j in range(m + 1):
        dp_table[0][j] = j

    # Fill the DP table using the recurrence relation
    for i in range(1, n + 1):
        for j in range(1, m + 1):
            # Cost of substitution: 0 if characters match, 1 otherwise
            cost_substitution = 0 if word1[i - 1] == word2[j - 1] else 1


            # Calculate the minimum of three possibilities:
            # 1. Deletion: dp_table[i-1][j] + 1 (delete character from word1)
            # 2. Insertion: dp_table[i][j-1] + 1 (insert character into word1 to match word2)
            # 3. Substitution: dp_table[i-1][j-1] + cost_substitution
            dp_table[i][j] = min(
                dp_table[i - 1][j] + 1,          # Deletion
                dp_table[i][j - 1] + 1,          # Insertion
                dp_table[i - 1][j - 1] + cost_substitution # Substitution
            )

    # The Levenshtein distance is the value at the bottom-right corner of the table
    distance = dp_table[n][m]

    return distance, dp_table

# --- Application to "easy" and "tease" ---
word_a = "easy"
word_b = "tease"

distance, table = levenshtein_distance_chart(word_a, word_b)

print(f"--- Levenshtein Distance Calculation for '{word_a}' and '{word_b}' ---")
print(f"Word 1: '{word_a}'")
print(f"Word 2: '{word_b}'")
print(f"\nCalculated Levenshtein Distance: {distance}")

print("\n--- Filled Dynamic Programming Table (Chart) ---")
# Print header for columns (word2 characters)
print("       " + " ".join([char for char in word_b]))
# Print rows
for i, row in enumerate(table):
    row_label = word_a[i-1] if i > 0 else " " # Label for word1 characters
    print(f"{row_label} [{i}] " + " ".join([str(val) for val in row]))

```
